```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("karkavelrajaj/amazon-sales-dataset")

print("Path to dataset files:", path)
```

```python
import os
os.listdir(path)
```

```python
df=pd.read_csv(path+"/amazon.csv")
df.head()
```

```python
df.info()
df.columns
```

```python
df['discounted_price'] = df['discounted_price'].astype(str)
df['actual_price'] = df['actual_price'].astype(str)
df['rating_count'] = df['rating_count'].astype(str)


#clean
df['discounted_price'] = df['discounted_price'].str.replace('₹','').str.replace(',','')
df['actual_price'] = df['actual_price'].str.replace('₹','').str.replace(',','')

df['discounted_price'] = pd.to_numeric(df['discounted_price'], errors='coerce')
df['actual_price'] = pd.to_numeric(df['actual_price'], errors='coerce')



df['rating'] = pd.to_numeric(df['rating'], errors='coerce')

df['rating_count'] = df['rating_count'].str.replace(',','')
df['rating_count'] = pd.to_numeric(df['rating_count'], errors='coerce')


#drop
df = df.dropna(subset=['actual_price', 'rating', 'rating_count'])
```

```python
import matplotlib.pyplot as plt
df['category'].value_counts().head(10).plot(kind='bar')
plt.title("Top Categories")
plt.show()
```

```python
#most expensive product
df.sort_values(by='actual_price', ascending=False).head(10)[['product_name','actual_price']]
```

```python
df['rating'].plot(kind='hist', bins=10)
plt.title("Ratings Distribution")
plt.show()
```

```python
plt.scatter(df['actual_price'], df['rating'])
plt.xlabel("Price")
plt.ylabel("Rating")
plt.title("Price vs Rating")
plt.show()
```

```python
#most reviewed product
df.sort_values(by='rating_count', ascending=False).head(10)[['product_name','rating_count']]
```

## Insights

- Electronics accessories such as HDMI cables and earphones dominate the dataset.
- A few products (like AmazonBasics HDMI cables and boAt earphones) have extremely high review counts, indicating strong popularity and demand.
- Products with very high ratings are not necessarily the most expensive, showing weak correlation between price and rating.
- The majority of products receive moderate to high ratings, indicating generally positive customer satisfaction.
- Smartphone-related products (e.g., Redmi series) also have high engagement, showing strong market demand in mobile devices.
