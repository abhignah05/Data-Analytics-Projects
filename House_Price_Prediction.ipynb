```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# 1: Encode and prepare data
df['Neighborhood_Quality'] = df['Neighborhood_Quality'].astype('float64')  # Assuming this is categorical
encoded_df = pd.get_dummies(df)

x = encoded_df.drop('House_Price', axis=1)
y = encoded_df['House_Price']

# Normalize
norm = MinMaxScaler().fit(x)
x = norm.transform(x)

# Train/test split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, shuffle=True)

# Train model (Before outlier removal)
lm = LinearRegression()
lm.fit(x_train, y_train)
y_pred_before = lm.predict(x_test)

# Evaluate model
mae_before = mean_absolute_error(y_test, y_pred_before)
rmse_before = np.sqrt(mean_squared_error(y_test, y_pred_before))

print("Before Outlier Removal:")
print("MAE:", mae_before)
print("RMSE:", rmse_before)

# Step 2: Remove outliers using IQR
def Outlier_Treatment(housedf, variable):
    percentile25 = housedf[variable].quantile(0.25)
    percentile75 = housedf[variable].quantile(0.75)
    iqr = percentile75 - percentile25
    upper_limit = percentile75 + 1.5 * iqr
    lower_limit = percentile25 - 1.5 * iqr
    housedf = housedf[(housedf[variable] >= lower_limit) & (housedf[variable] <= upper_limit)]
    return housedf

# Apply outlier removal
df_cleaned = Outlier_Treatment(df, 'House_Price')
df_cleaned = Outlier_Treatment(df_cleaned, 'Square_Footage')
df_cleaned = Outlier_Treatment(df_cleaned, 'Year_Built')
df_cleaned = Outlier_Treatment(df_cleaned, 'Lot_Size')

# Step 3: Repeat model after outlier removal
df_cleaned['Neighborhood_Quality'] = df_cleaned['Neighborhood_Quality'].astype('float64')
encoded_df_clean = pd.get_dummies(df_cleaned)

x_clean = encoded_df_clean.drop('House_Price', axis=1)
y_clean = encoded_df_clean['House_Price']

norm = MinMaxScaler().fit(x_clean)
x_clean = norm.transform(x_clean)

x_train_c, x_test_c, y_train_c, y_test_c = train_test_split(x_clean, y_clean, test_size=0.2, shuffle=True)

lm_clean = LinearRegression()
lm_clean.fit(x_train_c, y_train_c)
y_pred_after = lm_clean.predict(x_test_c)

# Evaluate cleaned model
mae_after = mean_absolute_error(y_test_c, y_pred_after)
rmse_after = np.sqrt(mean_squared_error(y_test_c, y_pred_after))

print("\nAfter Outlier Removal:")
print("MAE:", mae_after)
print("RMSE:", rmse_after)

# Comparison plot
plt.figure(figsize=(12,6))
plt.plot(y_test.values[:50], label='True Price (Before)', marker='o')
plt.plot(y_pred_before[:50], label='Predicted Price (Before)', marker='x')
plt.plot(y_test_c.values[:50], label='True Price (After)', marker='o')
plt.plot(y_pred_after[:50], label='Predicted Price (After)', marker='x')
plt.legend()
plt.title("Prediction Comparison Before and After Outlier Removal")
plt.xlabel("Sample")
plt.ylabel("House Price")
plt.grid(True)
plt.show()
```
